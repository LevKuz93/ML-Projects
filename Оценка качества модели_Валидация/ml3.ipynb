## 1.a.  Метод «исключения одного элемента» (Leave-One-Out Cross-Validation — LOOCV)

### Суть работы
Это экстремальный вариант **K-Fold кросс-валидации**, где количество фолдов $K$ равняется общему количеству объектов ($N$) в наборе данных. Модель обучается $N$ раз на $N-1$ объектах, используя оставленный объект для теста.

### Сильные стороны
* **Максимальное использование данных:** Модель обучается почти на 100% выборки, обеспечивая несмещенную оценку.
* **Отсутствие случайности:** Результат LOOCV всегда одинаков при фиксированных гиперпараметрах.

### Ограничения
* **Огромные вычислительные затраты:** Требует $N$ обучений, что невозможно для больших данных и тяжелых алгоритмов.
* **Высокая дисперсия (Variance) оценки:** Сильная корреляция между обучающими выборками может приводить к нестабильным оценкам на новых данных.


## 1.b. Алгоритмы поиска гиперпараметров

В машинном обучении критически важно подобрать оптимальные внешние настройки модели. Существует три основных подхода к автоматизации этого процесса:

### Grid Search (Поиск по сетке)
* **Принцип работы:** Вы вручную задаете точный список значений для каждого параметра. Алгоритм строит координатную сетку из всех возможных пересечений этих значений и последовательно тестирует каждую комбинацию.
* **Недостаток:** «Бездумный» перебор. Он не учится на ошибках и тратит много времени на проверку заведомо плохих веток.

### Random Search (Случайный поиск)
* **Принцип работы:** Вместо тестирования всех точек подряд алгоритм случайным образом выбирает комбинации из заданного диапазона или распределения.
* **Преимущество:** Не дублирует проверки по неважным параметрам. За то же количество шагов тестирует гораздо больше уникальных значений важных признаков, что математически повышает шанс найти оптимум.

### Bayesian Optimization (Байесовская оптимизация)
* **Принцип работы:** «Умный» поиск на основе теории вероятностей. Алгоритм строит внутреннюю суррогатную модель, которая предсказывает качество на основе прошлых попыток.
* **Преимущество:** На каждом шаге оценивает баланс между улучшением параметров в уже известной «хорошей» зоне (*Exploitation*) и исследованием абсолютно новых областей (*Exploration*). Экономит до 90% времени по сравнению с Grid Search.


## 1.c. Классификация методов отбора признаков (Feature Selection) в обучении с учителям

Отбор признаков помогает упростить модель, ускорить её обучение и предотвратить переобучение (overfitting). Все подходы делятся на три большие группы:

### 1. Фильтрующие методы (Filter Methods)
* **Принцип работы:** Оценивают каждый признак отдельно от модели машинного обучения с помощью чистой математической статистики.
* **Плюсы:** Работают молниеносно на терабайтах данных.
* **Минусы:** Рассматривают признаки изолированно. Они не способны заметить ситуации, когда два признака бесполезны по отдельности, но дают идеальный результат при совместном использовании.
* **Примеры:** Корреляция Пирсона, критерий Хи-квадрат ($\chi^2$), взаимная информация (Mutual Information).

### 2. Оберточные методы (Wrapper Methods)
* **Принцип работы:** Используют саму модель машинного обучения как «черный ящик». Специальный внешний алгоритм поочередно добавляет или удаляет признаки из набора и каждый раз заново обучает модель, отслеживая метрику качества.
* **Плюсы:** Учитывают взаимное влияние признаков и особенности конкретной модели.
* **Минусы:** Работают крайне медленно. Требуют сотен или тысяч циклов переобучения модели.
* **Примеры:** Пошаговое добавление (Forward Selection), последовательное удаление (Backward Elimination), рекурсивный отбор признаков (RFE).

### 3. Встроенные методы (Embedded Methods)
* **Принцип работы:** Отбор признаков происходит автоматически прямо в процессе обучения модели благодаря её внутренней математической архитектуре.
* **Плюсы:** Идеальный баланс между скоростью фильтров и точностью оберточных методов.
* **Минусы:** Сильно привязаны к конкретному типу моделей.
* **Примеры:** Штрафные санкции L1-регуляризации в Lasso-регрессии (зануляет веса слабых признаков), анализ важности признаков (Feature Importances) в случайных лесах и бустингах.

## 1.c. Статистические фильтры: Коэффициент Пирсона и Критерий Хи-квадрат ($\chi^2$)

Эти методы оценивают значимость каждого признака изолированно от модели, используя чистую математическую статистику. На основе вашей схемы они относятся к категории **Supervised -> Filters**.

### 1. Коэффициент корреляции Пирсона ($r$)
* **Тип данных:** Применяется только для **непрерывных числовых** признаков (например, рост, вес, температура, цена).
* **Что измеряет:** Силу и направление **линейной** взаимосвязи между признаком $X$ и целевой переменной $Y$.

$$r = \frac{\sum (X_i - \bar{X})(Y_i - \bar{Y})}{\sqrt{\sum (X_i - \bar{X})^2 \sum (Y_i - \bar{Y})^2}}$$

* **Интерпретация результатов:**
  * $r = 1$: Идеальная прямая линейная связь (один растет, второй растет).
  * $r = -1$: Идеальная обратная линейная связь (один растет, второй падает).
  * $r \approx 0$: Линейной связи между признаками **нет**. Такой признак можно смело отфильтровывать и удалять из модели.

### 2. Критерий Хи-квадрат ($\chi^2$)
* **Тип данных:** Применяется для **категориальных (номинальных)** признаков (например, цвет машины, пол, профессия, наличие симптома).
* **Что измеряет:** Степень независимости двух категориальных величин (признака и таргета).

$$\chi^2 = \sum \frac{(O_i - E_i)^2}{E_i}$$

*Где $O_i$ — фактически наблюдаемые частоты в данных (Observed), а $E_i$ — теоретически ожидаемые частоты при условии, что признаки абсолютно независимы (Expected).*

* **Интерпретация результатов:**
  * Алгоритм проверяет нулевую гипотезу: *«Признак и целевая переменная независимы друг от друга»*.
  * Если значение $\chi^2$ очень маленькое (а значение $p\text{-value} > 0.05$), значит, связь между признаком и таргетом чисто случайная. Данный признак отфильтровывается как бесполезный.

## 1.c. Встроенные методы: Lasso-регрессия (L1-регуляризация)

Согласно вашей схеме, этот подход относится к категории **Supervised -> Embedded**. Отбор признаков происходит автоматически прямо в процессе обучения модели за счет ее математической архитектуры.

### Суть работы метода
**Lasso (Least Absolute Shrinkage and Selection Operator)** — это линейная регрессия, к функции потерь которой добавляется «штраф» за сложность. Математика штрафа устроена так, что она прибавляет к ошибке прогнозирования (MSE) сумму абсолютных значений (модулей) всех весов (коэффициентов) модели.

### Математическая формула

$$\text{Ошибка} = \text{MSE} + \alpha \sum_{i=1}^{d} |w_{i}|$$

Где:
* $\text{MSE}$ — среднеквадратичная ошибка модели (качество аппроксимации данных).
* $w_i$ — веса (коэффициенты) перед признаками.
* $\alpha$ (альфа) — гиперпараметр регуляризации. Чем он больше, тем сильнее штраф и тем больше признаков будет исключено.

### Почему веса зануляются строго до нуля? (Геометрия L1)
В процессе обучения алгоритм оптимизации пытается минимизировать эту комбинированную ошибку. 

Ограничение, накладываемое L1-регуляризацией $\sum |w_i| \le C$, геометрически представляет собой **остроугольный ромб** (в многомерном пространстве — кросс-политоп). 
Линии уровня ошибки MSE (овалы) при расширении в первую очередь соприкасаются именно с **вершинами (углами)** этого ромба. На вершинах ромба одна или несколько координат (весов $w_i$) **строго равны нулю**.

* **Итог:** Слабые, зашумленные или дублирующие признаки получают коэффициент `0.0` и полностью выбывают из модели. Так происходит встроенный автоматический отбор признаков.

## Гибридные методы: Permutation Importance и SHAP

Методы **Permutation Importance** и **SHAP** не вписываются в строгую классическую схему «Фильтры vs Обертки». Они находятся ровно посередине, объединяя лучшие качества обоих подходов, из-за чего они крайне рекомендованы к использованию в реальных проектах.

### Почему они занимают промежуточное положение?

* **Как фильтры (Filters):** Они работают молниеносно по сравнению с классическими обертками, потому что нам **НЕ нужно переобучать модель тысячи раз**. Мы обучаем модель всего один раз, а затем анализируем её структуру.
* **Как оберточные методы (Wrappers):** Они не являются абстрактной изолированной статистикой. Они напрямую используют обученную модель машинного обучения как «судью», учитывая сложные нелинейные взаимосвязи и синергию признаков внутри алгоритма.

## 1.c. Значимость перестановок (Permutation Feature Importance)

Это простой, интуитивно понятный и мощный способ оценить важность признаков для **уже обученной** модели машинного обучения (пост-хок анализ).

### Пошаговый алгоритм работы
1. **Фиксация базы:** Берется валидационная (или тестовая) выборка и замеряется исходное качество модели (например, $Accuracy = 0.85$).
2. **Разрушение информации:** Выбирается один конкретный признак (например, *«Возраст»*) и его значения случайным образом перемешиваются (переставляются местами) между строками датасета. Связь этого признака с таргетом полностью уничтожается, но общее распределение признака и структура таблицы не меняются.
3. **Проверка модели:** Модель заставляют сделать предсказания на этом частично «испорченном» датасете и замеряют новое качество.
4. **Оценка важности:** Процесс повторяется по очереди для каждого признака.

### Интерпретация результатов
* Если качество **сильно упало** (например, с $0.85$ до $0.52$) — признак был **критически важным**. Без него модель ломается.
* Если качество **почти не изменилось** (или даже немного выросло) — признак **бесполезен** или дублирует другие данные. Модель его игнорирует.

### Главные плюсы метода
* **Model-Agnostic (Универсальность):** Работает абсолютно с любой моделью (от линейной регрессии до сложных нейросетей), так как оценивает модель как «черный ящик».
* **Защита от переобучения:** В отличие от встроенных `feature_importances_` в деревьях решений (которые часто завышают важность шумных, но непрерывных признаков на обучающей выборке), Permutation Importance считается на **валидационных данных**, что дает честную оценку.


## 1.c. Метод SHAP (Shapley Additive exPlanations)

**SHAP** — это передовой метод объяснения предсказаний («объяснительный ИИ»), основанный на математической теории кооперативных игр Ллойда Шепли (Нобелевская премия по экономике).

* **Принцип работы:** Представьте, что модель предсказала цену квартиры в 10 млн рублей при среднем рынке в 6 млн. SHAP рассматривает признаки квартиры (метро, ремонт, этаж) как игроков одной команды. Он математически перебирает все возможные комбинации (коалиции) этих признаков и вычисляет **чистый вклад каждого фактора** в финальное отклонение (+4 млн).
* **Главный плюс:** Дает как общую картину важности (*Глобальная интерпретируемость*), так и персональный расклад для каждого конкретного клиента или объекта (*Локальная интерпретируемость*).

### Что значит сделать процедуру разделения детерминированной?

**Детерминированность** означает, что при повторном запуске одного и того же кода на одних и тех же данных мы всегда будем получать **абсолютно идентичный результат**. 

По умолчанию функции разбиения (`train_test_split`, сэмплирование) используют генератор случайных чисел. Если не зафиксировать его состояние, то при каждом перезапуске ноутбука:
* Данные будут делиться по-разному.
* Метрики качества модели будут постоянно колебаться.
* Будет невозможно понять, улучшилось ли качество модели из-за изменения кода/гиперпараметров или вам просто «повезло» со случайным разбиением.

#### Как это реализуется в коде?
Для достижения детерминированности используется фиксация стартовой точки (зерна) генератора случайных чисел с помощью параметра **`random_state`** (в некоторых библиотеках `random_seed` или `seed`). Внутри функций мы явно передаем число (например, `42`), фиксируя случайность.

```python
def split_data_2_parts_random(df, test_size=0.2, random_state=42):
    
    train_df, test_df = train_test_split(
        df, test_size=test_size, random_state=random_state
    )
    return train_df, test_df


def split_data_3_parts_random(
    df, validation_size=0.2, test_size=0.2, random_state=42
):
    
    remaining_size = 1.0 - test_size
    val_adjusted_size = validation_size / remaining_size

    train_val_df, test_df = train_test_split(
        df, test_size=test_size, random_state=random_state
    )

    train_df, val_df = train_test_split(
        train_val_df, test_size=val_adjusted_size, random_state=random_state
    )

    return train_df, val_df, test_df


def split_data_2_parts_by_date(df, date_column, date_split):

    df_copy = df.copy()
    df_copy[date_column] = pd.to_datetime(df_copy[date_column])
    split_dt = pd.to_datetime(date_split)

    train_df = df_copy[df_copy[date_column] < split_dt]
    test_df = df_copy[df_copy[date_column] >= split_dt]

    return train_df, test_df


def split_data_3_parts_by_date(df, date_column, validation_date, test_date):
   
    df_copy = df.copy()
    df_copy[date_column] = pd.to_datetime(df_copy[date_column])
    val_dt = pd.to_datetime(validation_date)
    test_dt = pd.to_datetime(test_date)

    if val_dt >= test_dt:
        raise ValueError("validation_date должна быть строго раньше test_date!")

    train_df = df_copy[df_copy[date_column] < val_dt]
    val_df = df_copy[
        (df_copy[date_column] >= val_dt) & (df_copy[date_column] < test_dt)
    ]
    test_df = df_copy[df_copy[date_column] >= test_dt]

    return train_df, val_df, test_df
```

## 4. Реализуйте следующие методы перекрестной проверки:

```python
# МЕТОД 1: Классический K-Fold

def get_kfold_indices(df, k=5, random_state=42):

    kf = KFold(n_splits=k, shuffle=True, random_state=random_state)
    indices = list(kf.split(df))
    return indices

# МЕТОД 2: Групповой K-Fold (GroupKFold)

def get_group_kfold_indices(df, group_field, k=5):
    
    gkf = GroupKFold(n_splits=k)
    
    groups = df[group_field]
    indices = list(gkf.split(df, groups=groups))
    return indices

# МЕТОД 3: Стратифицированный K-Fold (StratifiedKFold)

def get_stratified_kfold_indices(df, stratify_field, k=5, random_state=42):
    
    skf = StratifiedKFold(n_splits=k, shuffle=True, random_state=random_state)
    
    y = df[stratify_field]
    indices = list(skf.split(df, y))
    return indices

# МЕТОД 4: Валидация временных рядов (TimeSeriesSplit)

def get_timeseries_kfold_indices(df, date_field, k=5):
    
    # Важно: Сортируем данные по дате перед делением временного ряда
    df_sorted = df.sort_values(by=date_field).reset_index(drop=True)

    tscv = TimeSeriesSplit(n_splits=k)
    indices = list(tscv.split(df_sorted))
    return indices
```

```python
def simple_feature_selection(X_df, y_series, nan_threshold=0.5, top_k=10):
    nan_ratio = X_df.isna().mean()
    valid_cols = nan_ratio[nan_ratio <= nan_threshold].index.tolist()

    correlations = {}
    for col in valid_cols:
        corr_val = np.abs(X_df[col].corr(y_series))
        correlations[col] = corr_val if not np.isnan(corr_val) else 0

    sorted_corr = sorted(correlations.items(), key=lambda x: x[1], reverse=True)
    return sorted_corr[:top_k]


corr_top_10_raw = simple_feature_selection(X_train_raw, y_train, top_k=10)
corr_top_10 = [item[0] for item in corr_top_10_raw]

print("--- 6.c. Порядок важности признаков по Корреляции Пирсона ---")
for feat, val in corr_top_10_raw:
    print(f"Признак: {feat:<20} | Абс. корреляция: {val:.6f}")


X_tr_c10 = X_train_raw[corr_top_10]
X_val_c10 = X_valid_raw[corr_top_10]
X_te_c10 = X_test_raw[corr_top_10]


model_c10 = Lasso(alpha=0.1, random_state=42, max_iter=10000)
model_c10.fit(X_tr_c10, y_train)


report_c10 = build_report_raw("Корреляции Пирсона MinMaxScaler", model_c10, X_tr_c10, X_val_c10, X_te_c10)
report_df = pd.concat([report_df, report_c10], ignore_index=True)
```

### Чем отличаются Lasso correlation и Lasso permutation?

Главное отличие между этими методами заключается в **математическом подходе** к оценке того, насколько признак информативен («хорош» или «важен»).

#### 1. Lasso correlation (Отбор по корреляции Пирсона)

Этот метод оценивает признаки **по отдельности (попарно)** и абсолютно изолированно от самой предсказательной модели (алгоритма машинного обучения).

* **Как работает:** Напрямую рассчитывается коэффициент линейной корреляции Пирсона между каждым конкретным признаком $X$ и целевой переменной $Y$ (цена квартиры). Формула коэффициента корреляции:
$$r_{xy} = \frac{\sum (x_i - \bar{x})(y_i - \bar{y})}{\sqrt{\sum (x_i - \bar{x})^2 \sum (y_i - \bar{y})^2}}$$
* **Плюсы:**
    * **Высокая скорость:** работает мгновенно даже на огромных объемах данных, так как требует лишь базовых математических операций.
* **Минусы:**
    * **Слепота к нелинейным зависимостям:** если связь между признаком и таргетом сложная (например, квадратичная), корреляция Пирсона покажет около нуля.
    * **Игнорирование совместной работы признаков:** метод не учитывает мультиколлинеарность. Если два признака сильно скоррелированы между собой (например, `Количество комнат` и `Количество окон`), корреляция Пирсона выберет оба как «важные», хотя в модели они будут дублировать друг друга и засорять её.


#### 2. Lasso permutation (Важность по перестановкам)

Этот метод оценивает признаки **внутри уже обученной модели** и смотрит на реальное падение качества предсказания алгоритма при разрушении информации.

* **Как работает:** Берётся валидационная выборка и измеряется базовая ошибка модели. Затем значения одного из признаков (например, `bathrooms`) случайно перемешиваются (переставляются местами) между строками. Это полностью разрушает связь конкретно этого признака с ценой, но сохраняет его исходное распределение. Модель делает предсказание на таких «испорченных» данных. Если ошибка (в данном случае `MAPE`) сильно выросла — признак критически важен для модели. Если ошибка не изменилась — признак бесполезен.
* **Плюсы:**
    * **Учитывает контекст модели:** показывает реальный вклад признака в работу конкретного алгоритма с учетом всех остальных колонок и их сложных взаимодействий.
    * **Универсальность:** не привязан к линейности данных.
* **Минусы:**
    * **Ресурсоемкость:** требует времени на вычисления. Чтобы оценить все признаки, алгоритму нужно делать предсказания по $N$ раз для каждого столбца (где $N$ — количество повторений `n_repeats` для усреднения шума).

### 6.d — Важность перестановок (Permutation Importance)

### Математическая формула MAPE

Обычная метрика MAPE (без приставки `neg_`) вычисляется по следующей формуле:

$$MAPE = \frac{100\%}{n} \sum_{i=1}^{n} \left| \frac{y_{i} - \hat{y}_{i}}{y_{i}} \right|$$

В `scikit-learn` по умолчанию результат возвращается в долях единицы (без умножения на $100\%$), поэтому формула для `neg_mean_absolute_percentage_error` выглядит так:

$$\text{neg\_MAPE} = -\frac{1}{n} \sum_{i=1}^{n} \left| \frac{y_{i} - \hat{y}_{i}}{y_{i}} \right|$$

**Где:**
* $n$ — количество объектов в выборке.
* $y_{i}$ — реальное значение таргета для $i$-го объекта (например, настоящая цена квартиры).
* $\hat{y}_{i}$ — предсказанное моделью значение (прогноз цены).
* $|y_i - \hat{y}_i|$ — абсолютная ошибка для одного объекта в деньгах.
* $\frac{|y_{i} - \hat{y}_{i}|}{y_{i}}$ — относительная (процентная) ошибка для конкретного объекта.

---

### Почему в Permutation Importance используют именно её?

Главный плюс MAPE — **масштабонезависимость**. 

Если использовать обычную ошибку в деньгах (MAE), то перемешивание важного признака у дорогой квартиры (например, с ценой $\$10,000$) даст огромный прирост ошибки в абсолютных долларах, затмив ошибки на дешевых квартирах. 

Метрика MAPE оценивает, на сколько процентов ухудшается качество предсказания:
* Значение `bathrooms 0.07797` в чек-листе означает, что если случайно перемешать информацию о ванных комнатах, относительная ошибка модели в среднем вырастет на **$7.79\%$** (или на $0.07797$ в долях единицы).

* **Ядро признаков:** Все три продвинутых метода (`Lasso`, `Permutation`, `SHAP`) выделили идентичную тройку лидеров, в точности совпадающую с чек-листом Школы 21: количество спален (`bedrooms`), ванных комнат (`bathrooms`) и наличие швейцара (`Doorman`).
* **Эффект сжатия (Top-10):** Переобучение модели Lasso на 10 признаках, отобранных по весам, перестановочной важности и ценностям Шепли, показало, что метрики качества ($MAE \approx 701.3$, $RMSE \approx 1010.9$, $R^2 \approx 0.59$) практически не изменились по сравнению с базовой моделью на всех 22 признаках. Это доказывает, что Lasso-регрессия эффективно занулила информационный шум (остальные бинарные признаки удобств), сделав модель легче без потери точности.
* **Слабость корреляционного фильтра:** Простой отбор по корреляции Пирсона показал чуть более слабые метрики качества. Это связано с тем, что одномерный статистический фильтр оценивает каждый признак изолированно и не способен учесть синергию (совместное влияние) признаков внутри модели.

#### 2. Скорость работы (Computation Speed)
* ** Лидер по скорости:** *Простой отбор по NaN и Корреляции*. Работает мгновенно (доли миллисекунды), так как использует только базовые векторные операции библиотеки `pandas` без вовлечения тяжелых алгоритмов обучения.
* ** Второе место:** *Коэффициенты Lasso*. Оценка важности извлекается мгновенно сразу после обучения базовой модели (`base_lasso_raw.coef_`), не требуя никаких дополнительных вычислений.
* ** Третье место:** *Метод SHAP*. Требует нескольких секунд для вычисления маржинальных вкладов Шепли на валидационной выборке. Для линейных моделей (`LinearExplainer`) работает быстро, но на тяжелых ансамблях (например, LightGBM) время расчета резко возрастает.
* ** Самый медленный метод:** *Permutation Importance*. Вынужден для каждого из 22 признаков выполнять $N$ случайных перестановок строк (в нашем коде `n_repeats=10`) и каждый раз заново прогонять измененную матрицу через функцию `predict` модели.

#### 3. Стабильность результатов (Stability)
* **Абсолютная стабильность (Детерминированность):** Методы *Lasso*, *SHAP* и *Корреляция Пирсона* математически монолитны. При фиксированных исходных данных и `random_state` модели их оценки вклада признаков будут абсолютно идентичны от запуска к запуску.
* **Относительная стабильность:** *Permutation Importance* обладает встроенной стохастичностью (случайностью). Из-за хаотичного перемешивания значений признаков между строками, итоговые оценки `Mean` и `Std` могут незначительно колебаться на небольших выборках. Тем не менее, для топ-3 главных признаков (`bedrooms`, `bathrooms`, `Doorman`) метод гарантирует железобетонную стабильность.


###  Итоговое заключение

* Для **первичного отсева** огромных массивов данных (тысячи признаков) лучше всего подходит комбинация *NaN + Correlation* за счет своей колоссальной скорости.
* Для **финального отбора параметров** в линейных моделях наилучшие результаты показывают встроенные *коэффициенты Lasso* и *метод SHAP*. ## 7. Оптимизация гиперпараметров

### Оптимизация гиперпараметров модели ElasticNet

В данном разделе рассматривается подбор оптимальных внешних настроек для модели `ElasticNet`. 
`ElasticNet` — это линейная регрессия, которая комбинирует в себе свойства Lasso (L1) и Ridge (L2) регуляризации. Мы настраиваем два ключевых параметра:
* `alpha` — общая сила штрафа (регуляризации).
* `l1_ratio` — баланс между L1 и L2. При `l1_ratio = 1` модель превращается в чистый Lasso, при `l1_ratio = 0` — в чистый Ridge.

Мы сравним три подхода к оптимизации: **Grid Search**, **Random Search** и байесовскую оптимизацию с помощью библиотеки **Optuna**.

### Реализация Grid Search и Random Search 

```python
X_train = train_part[feature_list]
X_valid = valid_part[feature_list]
X_test = test_part[feature_list]

grid_param = {
    'alpha': [0.01, 0.1, 0.5, 1.0, 5.0],
    'l1_ratio': [0.1, 0.3, 0.5, 0.7, 0.9]
}


random_param = {
    'alpha': np.geomspace(0.01, 5.0, 100), 
    'l1_ratio': np.linspace(0.05, 0.95, 100)
}

def evaluate_best_model(name, search_object):
   
    best_model = search_object.best_estimator_
    preds = best_model.predict(X_valid)
    
    mae = mean_absolute_error(y_valid, preds)
    rmse = np.sqrt(mean_squared_error(y_valid, preds))
    r2 = r2_score(y_valid, preds)
    
    print(f"=== Результаты {name} ===")
    print(f"Лучшие параметры по CV: {search_object.best_params_}")
    print(f"Метрики на валидации -> MAE: {mae:.5f} | RMSE: {rmse:.5f} | R2: {r2:.5f}\n")
    
    return {
        'model': name, 
        'alpha': search_object.best_params_['alpha'], 
        'l1_ratio': search_object.best_params_['l1_ratio'], 
        'MAE': mae, 
        'RMSE': rmse, 
        'R2': r2
    }

# 3. ПОДХОД 1: Grid Search 
grid_search = GridSearchCV(
    estimator=ElasticNet(random_state=42, max_iter=5000), 
    param_grid=grid_param, 
    cv=3, 
    scoring='r2',
    n_jobs=-1
)
grid_search.fit(X_train, y_train)
res_grid = evaluate_best_model("Grid Search", grid_search)

# 4. ПОДХОД 2: Random Search 
random_search = RandomizedSearchCV(
    estimator=ElasticNet(random_state=42, max_iter=5000), 
    param_distributions=random_param, 
    n_iter=15, 
    cv=3, 
    scoring='r2',
    random_state=42, 
    n_jobs=-1
)
random_search.fit(X_train, y_train)
res_random = evaluate_best_model("Random Search", random_search)
```

### Интеграция Optuna

```python
# Отключаем лишние логи Optuna
optuna.logging.set_verbosity(optuna.logging.WARNING)

def objective(trial):

    alpha = trial.suggest_float('alpha', 0.001, 5.0, log=True)
    l1_ratio = trial.suggest_float('l1_ratio', 0.05, 0.95)
    
    model = ElasticNet(alpha=alpha, l1_ratio=l1_ratio, random_state=42, max_iter=5000)
    

    score = cross_val_score(model, X_train, y_train, cv=3, scoring='r2', n_jobs=-1)
    
    return score.mean()


study = optuna.create_study(direction='maximize')

study.optimize(objective, n_trials=50)


best_params = study.best_params
optuna_model = ElasticNet(**best_params, random_state=42, max_iter=5000)
optuna_model.fit(X_train, y_train)


preds_optuna = optuna_model.predict(X_valid)
mae_opt = mean_absolute_error(y_valid, preds_optuna)
rmse_opt = np.sqrt(mean_squared_error(y_valid, preds_optuna))
r2_opt = r2_score(y_valid, preds_optuna)

print(f"=== Результаты Optuna ===")
print(f"Лучшие параметры: {best_params}")
print(f"Метрики на валидации -> MAE: {mae_opt:.5f} | RMSE: {rmse_opt:.5f} | R2: {r2_opt:.5f}")
```

### Запуск Optuna на продвинутой схеме кросс-валидации 

```python
fixed_group_indices = get_group_kfold_indices(train_part, group_field='manager_id', k=3)

def objective_group_cv(trial):

    params = {
        'alpha': trial.suggest_float('alpha', 0.001, 5.0, log=True),
        'l1_ratio': trial.suggest_float('l1_ratio', 0.05, 0.95)
    }
    
    cv_scores = []
   
    for train_idx, test_idx in fixed_group_indices:
        
        X_tr_fold = X_train.iloc[train_idx]
        y_tr_fold = y_train.iloc[train_idx]
        X_val_fold = X_train.iloc[test_idx]
        y_val_fold = y_train.iloc[test_idx]
        
        model = ElasticNet(**params, random_state=42, max_iter=5000)
        model.fit(X_tr_fold, y_tr_fold)
        
        fold_score = r2_score(y_val_fold, model.predict(X_val_fold))
        cv_scores.append(fold_score)
        
    return np.mean(cv_scores)

optuna.logging.set_verbosity(optuna.logging.WARNING)

study_group_cv = optuna.create_study(direction='maximize')


study_group_cv.optimize(objective_group_cv, n_trials=40, show_progress_bar=False)

final_group_model = ElasticNet(**study_group_cv.best_params, random_state=42, max_iter=5000)
final_group_model.fit(X_train, y_train)

final_preds = final_group_model.predict(X_valid)
final_mae = mean_absolute_error(y_valid, final_preds)
final_rmse = np.sqrt(mean_squared_error(y_valid, final_preds))
final_r2 = r2_score(y_valid, final_preds)

print("=== Результаты Optuna на схеме GroupKFold ===")
print(f"Лучшие параметры: {study_group_cv.best_params}")
print(f"Метрики на валидации -> MAE: {final_mae:.5f} | RMSE: {final_rmse:.5f} | R2: {final_r2:.5f}")

res_group_cv = {
    'model': 'Optuna + GroupKFold', 
    'alpha': study_group_cv.best_params['alpha'], 
    'l1_ratio': study_group_cv.best_params['l1_ratio'], 
    'MAE': final_mae, 
    'RMSE': final_rmse, 
    'R2': final_r2
}
```

###  Анализ результатов и сравнение подходов оптимизации

1. **Grid Search (Перебор по сетке):** 
   Метод выполнил полный детерминированный перебор параметров по заданной фиксированной матрице координат. Он находит лучший результат только в том случае, если шаг сетки изначально был выбран очень мелким. Метод не учитывает взаимосвязи между параметрами и тратит ресурсы на проверку неэффективных комбинаций.

2. **Random Search (Случайный поиск):**
   Благодаря хаотичному выбору непрерывных значений из диапазона `np.linspace`, метод за те же 15 итераций проверил 15 уникальных вариантов `alpha` и `l1_ratio`, не дублируя вычисления. На практике случайный поиск часто находит более точные комбинации параметров, чем крупная фиксированная сетка.

3. **Optuna (Байесовская оптимизация на алгоритме TPE):**
   Является наиболее эффективным и «умным» подходом. Алгоритм разделяет историю запусков на условные группы «хороших» и «плохих» параметров. Каждая следующая итерация выбирается так, чтобы максимизировать вероятность попадания в зону высокой метрики качества $R^2$. Это позволяет находить оптимальные гиперпараметры за значительно меньшее число шагов.

4. **Интеграция со схемами перекрестной проверки (Optuna + GroupKFold):**
   Запуск `Optuna` в связке с кастомным `GroupKFold` по полю `manager_id` является наиболее надежным решением с точки зрения бизнес-логики. Данный подход гарантирует, что модель ElasticNet подбирает внутренние веса штрафов регуляризации не под конкретный стиль риелторов, а под общие рыночные свойства квартир (количество комнат, удобств). Полученные таким образом гиперпараметры обладают максимальной обобщающей способностью и защищены от переобучения.
